1. Memuat Data dan Melihat Distribusi Kelas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data yang sudah dibersihkan
df = pd.read_csv('student_data_cleaned.csv')

# Visualisasi Distribusi Kelas Target
plt.figure(figsize=(7, 5))
ax = sns.countplot(x='passed', data=df, palette='Set2')
plt.title('Distribusi Status Kelulusan Siswa', fontsize=14, pad=15)
plt.xlabel('Status (0 = Berisiko, 1 = Aman)', fontsize=12)
plt.ylabel('Jumlah Siswa', fontsize=12)

# Menambahkan angka di atas batang
for p in ax.patches:
    ax.annotate(f'{p.get_height()}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 10), textcoords='offset points')
plt.show()
```

2. Mencari Faktor yang Paling Mempengaruhi Kelulusan

```python
plt.figure(figsize=(8, 10))
korelasi = df.corr()[['passed']].sort_values(by='passed', ascending=False)
korelasi_top10 = korelasi.drop('passed').head(10)
korelasi_terbawah = korelasi.tail(5)


korelasi_penting = pd.concat([korelasi_top10, korelasi_terbawah])

sns.heatmap(korelasi_penting, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f")
plt.title('Faktor Paling Berpengaruh terhadap Kelulusan', fontsize=14, pad=15)
plt.show()
```

3. Membedah Faktor Dominan

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='passed', y='academic_habit_score', data=df, palette='pastel')
plt.title('Perbandingan Skor Kebiasaan Akademik: Aman vs Berisiko', fontsize=14, pad=15)
plt.xlabel('Status (0 = Berisiko, 1 = Aman)', fontsize=12)
plt.ylabel('Academic Habit Score (Scaled)', fontsize=12)
plt.show()
```
